## Build DataFrames

```python
# ----- SERIES -----
series_list = []
series_map = {}  # to map series_name → series_id

for mt in data.get('typeMatches', []):
    for series in mt.get('seriesMatches', []):
        wrapper = series.get('seriesAdWrapper')
        if wrapper:
            series_id = wrapper.get("seriesId")
            series_name = wrapper.get("seriesName")
            if series_id not in series_map:
                series_map[series_name] = series_id
                series_list.append({
                    "series_id": series_id,
                    "series_name": series_name,
                    "start_date": startDate,  # optional
                    "end_date": endDate,
                    "tournament_Format": matchFormat,
                    "tournament_type": matchType
                })

df_series = pd.DataFrame(series_list)

# ----- VENUES -----
venue_map = {}  # ground → venue_id
venues_list = []

matches_list = []
teams_list = []
team_map = {}  # team_name → team_id
team_id_counter = 1

for mt in data.get('typeMatches', []):
    for series in mt.get('seriesMatches', []):
        wrapper = series.get('seriesAdWrapper')
        if wrapper:
            for match in wrapper.get('matches', []):
                info = match['matchInfo']
                
                # Teams
                for t in ['team1', 'team2']:
                    team_name = info[t]['teamName']
                    if team_name not in team_map:
                        team_map[team_name] = team_id_counter
                        teams_list.append({
                            "team_id": team_id_counter,
                            "team_name": team_name,
                            "abbreviation": info[t].get('teamSName', None)
                        })
                        team_id_counter += 1
                
                # Venue
                venue = info.get('venueInfo', {})
                ground = venue.get('ground')
                if ground and ground not in venue_map:
                    venue_map[ground] = len(venue_map)+1
                    venues_list.append({
                        "venue_id": venue_map[ground],
                        "ground": ground,
                        "city": venue.get('city'),
                        "country": venue.get('country')
                    })
                
                # Matches
                matches_list.append({
                    "match_id": info.get("matchId"),
                    "team1_id": team_map[info['team1']['teamName']],
                    "team2_id": team_map[info['team2']['teamName']],
                    "series_id": wrapper.get("seriesId"),
                    "venue_id": venue_map.get(ground),
                    "status": info.get("status"),
                    "format": info.get("matchFormat"),
                    "start_date": None
                })

df_teams = pd.DataFrame(teams_list)
df_venues = pd.DataFrame(venues_list)
df_matches = pd.DataFrame(matches_list)
```

```python
import requests

API_KEY = "c052ee089bmshd7550087d8cb28ap196360jsn97d96515789d"
headers = {
    "X-RapidAPI-Key": API_KEY,
    "X-RapidAPI-Host": "cricbuzz-cricket.p.rapidapi.com"
}

# Fetch recent matches (or specify format via filtering)
url = "https://cricbuzz-cricket.p.rapidapi.com/matches/v1/recent"
res = requests.get(url, headers=headers)
matches = res.json()
```

```python
matches_scores = []

for fmt in matches.get('typeMatches', []):  # Test, ODI, T20I
    for series in fmt.get('seriesMatches', []):
        matches_list = series.get('seriesAdWrapper', {}).get('matches', [])
        for m in matches_list:
            match_info = m.get('matchInfo', {})
            if not match_info:
                continue

            match_id = match_info.get('matchId')
            match_format = match_info.get('matchFormat')

            # Fetch scorecard
            sc_url = f"https://cricbuzz-cricket.p.rapidapi.com/mcenter/v1/{match_id}/scard"
            sc_res = requests.get(sc_url, headers=headers)
            sc_data = sc_res.json()

            # Extract player innings
            for inning in sc_data.get("scorecard", []):
                for b in inning.get("batsman", []):
                    matches_scores.append({
                        "player": b.get("name"),
                        "runs": b.get("runs"),
                        "format": match_format
                    })
```

```python
matches_scores
```

```python
highest = {}
for entry in matches_scores:
    fmt = entry['format']
    if fmt not in highest or entry['runs'] > highest[fmt]['runs']:
        highest[fmt] = entry

for fmt, data in highest.items():
    print(f"{fmt}: {data['player']} - {data['runs']} runs")
```

## Step 2: Get Players from Match (Playing XI)

```python
import requests

match_id = "149743"

headers = {
    "X-RapidAPI-Key": "c052ee089bmshd7550087d8cb28ap196360jsn97d96515789d",
    "X-RapidAPI-Host": "cricbuzz-cricket.p.rapidapi.com"
}

url = f"https://cricbuzz-cricket.p.rapidapi.com/mcenter/v1/{match_id}/scard"

res = requests.get(url, headers=headers)
data = res.json()

players = []

for inning in data.get("scoreCard", []):
    
    # Batters
    batsmen = inning.get("batTeamDetails", {}).get("batsmenData", {})
    for b in batsmen.values():
        players.append((b.get("batId"), b.get("batName")))
    
    # Bowlers
    bowlers = inning.get("bowlTeamDetails", {}).get("bowlersData", {})
    for bw in bowlers.values():
        players.append((bw.get("bowlerId"), bw.get("bowlName")))

# Remove duplicates
players = list(set(players))

print(players)
```

```python
players = []

for inning in data.get("scorecard", []):
    
    # Batters
    for b in inning.get("batsman", []):
        players.append((b.get("id"), b.get("name")))
    
    # Bowlers (IMPORTANT)
    for bw in inning.get("bowler", []):
        players.append((bw.get("id"), bw.get("name")))

# Remove duplicates
players = list(set(players))

print(players)
```
