```python
import pandas as pd

def flatten_match(data):
    matches = data.get("typeMatches", [])

    rows = []

    for mt in matches:
        match_type = mt.get("matchType")

        for series in mt.get("seriesMatches", []):
            series_info = series.get("seriesAdWrapper", {})

            for match in series_info.get("matches", []):

                m = match.get("matchInfo", {})
                score = match.get("matchScore", {})

                row = {
                    # 🏏 match info
                    "match_id": m.get("matchId"),
                    "series_id": m.get("seriesId"),
                    "series_name": m.get("seriesName"),
                    "match_desc": m.get("matchDesc"),
                    "match_format": m.get("matchFormat"),
                    "start_date": datetime.fromtimestamp(int(m.get("startDate"))/1000),
                    "end_date": datetime.fromtimestamp(int(m.get("endDate"))/1000),
                    "state": m.get("state"),
                    "status": m.get("status"),

                    # 🏟️ teams
                    "team1_id": m.get("team1", {}).get("teamId"),
                    "team1_name": m.get("team1", {}).get("teamName"),
                    "team1_short": m.get("team1", {}).get("teamSName"),

                    "team2_id": m.get("team2", {}).get("teamId"),
                    "team2_name": m.get("team2", {}).get("teamName"),
                    "team2_short": m.get("team2", {}).get("teamSName"),

                    # 📊 scores
                    "team1_runs": score.get("team1Score", {}).get("inngs1", {}).get("runs"),
                    "team1_wkts": score.get("team1Score", {}).get("inngs1", {}).get("wickets"),
                    "team1_overs": score.get("team1Score", {}).get("inngs1", {}).get("overs"),

                    "team2_runs": score.get("team2Score", {}).get("inngs1", {}).get("runs"),
                    "team2_wkts": score.get("team2Score", {}).get("inngs1", {}).get("wickets"),
                    "team2_overs": score.get("team2Score", {}).get("inngs1", {}).get("overs"),

                    # 📍 venue
                    "venue_id": m.get("venueInfo", {}).get("id"),
                    "ground": m.get("venueInfo", {}).get("ground"),
                    "city": m.get("venueInfo", {}).get("city"),

                    # 🧠 result
                    "result": m.get("status")
                }

                rows.append(row)

    return pd.DataFrame(rows)
```

```python
df = flatten_match(data)
```

## Match_info

```python
import requests
import pandas as pd

match_records = []
player_records = []

match_ids = df["match_id"]

for mid in match_ids:
    url = f"https://cricbuzz-cricket.p.rapidapi.com/mcenter/v1/{mid}"
    response = requests.get(url, headers=headers)
    data = response.json()

    match = data.get("matchInfo", {})

    # 🔹 Match-level data
    match_records.append({
        "match_id": match.get("matchId"),
        "description": match.get("matchDescription"),
        "format": match.get("matchFormat"),
        "type": match.get("matchType"),
        "status": match.get("status"),
        "complete": match.get("complete"),
        "start_time": match.get("matchStartTimestamp"),
        "end_time": match.get("matchCompleteTimestamp"),
        "team1": match.get("team1", {}).get("name"),
        "team2": match.get("team2", {}).get("name"),
        "series": match.get("series", {}).get("name"),
        "venue": match.get("venue", {}).get("name"),
        "city": match.get("venue", {}).get("city"),
        "winner": match.get("result", {}).get("winningTeam"),
        "margin": match.get("result", {}).get("winningMargin")
    })

    # 🔹 Player-level data
    for team_key in ["team1", "team2"]:
        team = match.get(team_key, {})

        for p in team.get("playerDetails", []):
            player_records.append({
                "match_id": match.get("matchId"),
                "team_id": team.get("id"),
                "team_name": team.get("name"),
                "player_id": p.get("id"),
                "name": p.get("name"),
                "full_name": p.get("fullName"),
                "role": p.get("role"),
                "batting_style": p.get("battingStyle"),
                "bowling_style": p.get("bowlingStyle"),
                "captain": p.get("captain"),
                "keeper": p.get("keeper"),
                "substitute": p.get("substitute")
            })
    toss = match.get("tossResults", {})
    result = match.get("result", {})
    
    match_records.append({
        "match_id": match.get("matchId"),
        "team1": match.get("team1", {}).get("name"),
        "team2": match.get("team2", {}).get("name"),
        
        # 🔹 Toss
        "toss_winner": toss.get("tossWinnerName"),
        "toss_winner_id": toss.get("tossWinnerId"),
        "toss_decision": toss.get("decision"),
    
        # 🔹 Result
        "winner": result.get("winningTeam"),
        "winner_id": result.get("winningteamId"),
        "margin": result.get("winningMargin"),
        "win_by_runs": result.get("winByRuns"),
        "win_by_innings": result.get("winByInnings"),
        "result_type": result.get("resultType")
    })

# 🔹 Create final DataFrames
match_df = pd.DataFrame(match_records)
players_df = pd.DataFrame(player_records)
```
